```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Amit", "Rahul", "Priya", "Sneha", "Rohan"],
    "Department": ["CE", "IT", "CE", "IT", "CE"],
    "Marks": [85, np.nan, 78, 92, np.nan],
    "Age": [20, 21, 19, np.nan, 22]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# Detect missing values
print("\nMissing values:")
print(df.isnull())

print("\nNumber of missing values:")
print(df.isnull().sum())

# Fill missing values
df_filled = df.copy()

df_filled["Marks"] = df_filled["Marks"].fillna(
    df_filled["Marks"].mean()
)

df_filled["Age"] = df_filled["Age"].fillna(
    df_filled["Age"].mean()
)

print("\nAfter filling missing values:")
print(df_filled)

# Drop rows containing missing values
df_dropped = df.dropna()

print("\nAfter dropping missing values:")
print(df_dropped)

# Filtering rows
print("\nStudents with marks greater than 80:")
print(df_filled[df_filled["Marks"] > 80])

# Selecting columns
print("\nName and Marks:")
print(df_filled[["Name", "Marks"]])

# Sorting
print("\nSorted by Marks:")
print(df_filled.sort_values("Marks", ascending=False))

# Grouping
print("\nAverage marks by department:")
print(df_filled.groupby("Department")["Marks"].mean())
```
