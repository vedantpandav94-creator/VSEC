```python

import pandas as pd

# Read CSV
df = pd.read_csv("data.csv")

# 1. Check missing values
print(df.isnull())

# Count missing values in each column
print(df.isnull().sum())


# 2. Fill missing values
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Salary"] = df["Salary"].fillna(0)

print(df)


# 3. Drop rows containing missing values
df_clean = df.dropna()

print(df_clean)


# 4. Replace data
df["City"] = df["City"].replace("Pune", "Pimpri")

print(df)


# 5. Filter rows
# Employees whose salary is greater than 35000
filtered_rows = df[df["Salary"] > 35000]

print(filtered_rows)


# 6. Select/filter columns
selected_columns = df[["Name", "City", "Salary"]]

print(selected_columns)


# 7. Sort values
# Sort by salary from highest to lowest
sorted_df = df.sort_values("Salary", ascending=False)

print(sorted_df)


# 8. Group values
# Find average salary for each department
grouped = df.groupby("Department")["Salary"].mean()

print(grouped)


# 9. Save final data to Excel
df.to_excel("output.xlsx", index=False)

print("Excel file created successfully!")
```
