```python

import seaborn as sns
import matplotlib.pyplot as plt 
```

```python
c = [1, 2, 3,4,5]
v = [2, 4, 6,8,10]
sns.lineplot(x=c, y=v)
plt.title("Line Plot")
plt.show()  
```

```python
c = [1, 2, 3,4,5]
v = [2, 4, 6,8,10]
sns.barplot(x=c, y=v)
plt.title("Line Plot")
plt.show()
```

```python
c = [1, 2, 3,4,5]
v = [2, 4, 6,8,10]
sns.histplot(x=c, y=v)
plt.title("Line Plot")
plt.show()
```

```python
c = [1, 2, 3,4,5]
v = [2, 4, 6,8,10]
sns.boxplot(x=c, y=v)
plt.title("Line Plot")
plt.show()
```

```python
c = [1, 2, 3,4,5]
v = [2, 4, 6,8,10]
sns.scatterplot(x=c, y=v)
plt.title("Line Plot")
plt.show()   
```
