```python
import numpy as np

# Random numbers between 0 and 1
random_numbers = np.random.rand(5)
print("Random numbers:")
print(random_numbers)

# Standard normal distribution
normal_numbers = np.random.randn(5)
print("\nStandard normal random numbers:")
print(normal_numbers)

# Random integers
integers = np.random.randint(1, 101, 10)
print("\nRandom integers:")
print(integers)

# Normal distribution
normal_data = np.random.normal(
    loc=50,
    scale=10,
    size=1000
)

print("\nNormal distribution data:")
print(normal_data[:10])

# Statistics
print("\nMean:", np.mean(normal_data))
print("Median:", np.median(normal_data))
print("Standard deviation:", np.std(normal_data))
print("Variance:", np.var(normal_data))
print("Minimum:", np.min(normal_data))
print("Maximum:", np.max(normal_data))
```
