```python
import matplotlib.pyplot as plt

# Example data
x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 5, 6]

cats = ["A", "B", "C", "D"]
dog = [10, 15, 7, 12]

data = [1, 2, 2, 3, 3, 3, 4, 4, 5, 6, 6, 7]

plt.subplot(2, 2, 1)
plt.plot(x, y)
plt.title("Line Plot")

plt.subplot(2, 2, 2)
plt.bar(cats, dog)
plt.title("Bar Chart")

plt.subplot(2, 2, 3)
plt.scatter(x, y)
plt.title("Scatter Plot")

plt.subplot(2, 2, 4)
plt.hist(data)
plt.title("Histogram")

plt.tight_layout()
plt.show()
```
